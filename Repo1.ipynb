# **Introducción**

En Chile y el mundo los niños, niñas y adolescentes son fundamentales para el desarrollo social, económico y cultural. Sin embargo, no es común hablar sobre el maltrato infantil, ya que muchas veces se percibe como un tema tabú o como algo perteneciente a problema del pasado. No obstante, en la actualidad continúan registrándose casos de abuso y violencia, tanto fuera como dentro de sus hogares.

Las consecuencias de estos actos tienen un impacto profundo en el desarrollo de los menores, afectando diferentes ámbitos como su salud física y mental, el rendimiento académico y las relaciones sociales. Estos efectos pueden generar, a largo plazo, dificultades emocionales, conductuales y sociales que repercuten en su bienestar y en su integración en la sociedad.

Por ello, es crucial visibilizar estos actos contra la niñez y abordar esta problemática de manera integral. Reconocer y analizar la magnitud de la violencia infantil permite diseñar estrategias de prevención, protección y apoyo, contribuyendo a garantizar un entorno seguro y propicio para el desarrollo de todos los niños y adolescentes.

# **Contexto y Motivación**

La niñez representa una etapa fundamental en el desarrollo humano, ya que marca diversos
procesos físicos, cognitivos y emocionales. Este período es sumamente importante, pues
durante él se desarrollan habilidades, valores y competencias que sientan las bases para la
vida adulta. La protección de los derechos infantiles y el establecimiento de relaciones
afectivas seguras y consistentes son esenciales, ya que los vínculos saludables, basados en
el cariño y en límites claros, favorecen la confianza, la autoestima y la capacidad de
relacionarse de manera positiva con los demás. Estos pilares permiten al niño desarrollar
seguridad emocional y social, creando una base sólida para enfrentar los desafíos de la vida
adulta.

Actualmente, en Chile existen 4.493.995 niños, niñas y adolescentes, lo que representa
aproximadamente el 24,3% de la población total del país. Este grupo de la población posee
derechos fundamentales que buscan garantizar su desarrollo integral, incluyendo el derecho
a la vida, educación, salud, libertad de expresión y a la protección frente a toda forma de
violencia, entre otros. No obstante, en la práctica, estos derechos no siempre se cumplen
plenamente, lo que puede generar impactos negativos en su bienestar físico, emocional y
social, limitando sus oportunidades de crecimiento y desarrollo pleno.

La violencia infantil constituye un problema tanto social como de salud pública, ya que genera
consecuencias profundas en el desarrollo emocional, social y educativo de quienes la sufren,
tanto a corto como a largo plazo. Los niños y adolescentes expuestos a violencia pueden
presentar dificultades de aprendizaje, problemas de conducta, afectación en sus relaciones
interpersonales y efectos negativos duraderos en su bienestar psicológico.

Es por esta razón que nuestra motivación se centra en comprender y evidenciar la existencia
de violencia hacia niños, niñas y adolescentes en Chile. Queremos destacar cómo los
distintos tipos de violencia afectan el bienestar emocional de quienes los sufren, permitiendo
identificar diferentes patrones, tendencias y cambios que surgen a lo largo del tiempo. Por
ello, nuestro público objetivo incluye diversos órganos gubernamentales, centros de salud,
establecimientos educacionales, familias y círculos cercanos, fuerzas de seguridad y
entidades dedicadas a la protección infantil, entre otros centros de interés. El objetivo es
generar conciencia sobre estas vulneraciones, con el fin de promover políticas, intervenciones
y acciones concretas que contribuyan a erradicar la violencia, garantizando un cuidado digno
y un desarrollo seguro para todos los niños, niñas y adolescentes de Chile

### ***- ¿Qué tipos de agresión (física, emocional, sexual) son los más frecuentes expuestos en los datos?***

### - **Fisicas**

```python
col = ["PA_5", "PA_6", "PA_7", "PB_2", "PC_1", "PC_2", "PC_5"]
fisica = df_fisica[df_fisica[col].eq(1).any(axis=1)]
```

### - **Verbal**

```python
col = ['PB_1', 'PC_3']
verbal = df_verbal[df_verbal[col].eq(1).any(axis=1)]
```

### - **Sexual**

```python
col = ["PD_3", "PD_4", "PD_5", "PD_6", "PD_7"]
sexual = df_sexual[df_sexual[col].eq(1).any(axis=1)]
```

```python
# Contamos los casos según el tipo de violencia por cada año 
conteo_F = fisica["Año"].value_counts().sort_index()
conteo_V = verbal["Año"].value_counts().sort_index()
conteo_S = sexual["Año"].value_counts().sort_index()

df_cant = pd.DataFrame({"Física": conteo_F, "Verbal": conteo_V, "Sexual": conteo_S}).fillna(0).astype(int)
df_cant
```

```python
# Graficamos
df_cant.plot(kind='bar', color=['#95785a', '#979b44', '#dfdab0']) 

plt.title("\n Tipos de agresión por año\n",fontsize=18, fontweight='bold')
plt.xlabel("\nAño", fontsize= 12, color='grey')
plt.ylabel("Cantidad de personas\n", fontsize=12, color='grey')

plt.xticks(rotation=0)
plt.legend(title="Tipo de agresión")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### ***-¿Cuál es la proporción de casos de violencia física atribuida a pares (otros niños, niñas o adolescentes), personas cercanas, conocidos y extraños dentro de los datos?***

```python
violencia_fisica = ["PA_5_4","PA_6_4","PA_6_5","PA_7_4","PA_7_5","PB_2","PB_2_4","PC_1","PC_2","PC_5","PC_5_3"]
df_inicial[violencia_fisica] = df_inicial[violencia_fisica].replace(0, pd.NA)
df_inicial = df_inicial[df_inicial[violencia_fisica].notna().any(axis=1)]
```

```python
def clasificar_agresor(row):
    if (
       (pd.notna(row.get("PA_5_4")) and row["PA_5_4"] in [3,4])  # "PA_5_4 = [3,4] indica que sea adolescente, niño o niña.
       or (pd.notna(row.get("PC_1")) and row["PC_1"] == 1)       # PC_1 = 1 especifica que fue agredido por un niño, niña o adolescente.
       or (pd.notna(row.get("PC_2")) and row["PC_2"] == 1)       # PC_2 = 1 especifica que fue agredido por un grupo de niños, niñas o adolescentes.
    ): 
        return "Pares"

    if (
       (pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 1)      # PA_6_4 = 1 indica que fue agredido por alguien cercano con quien vive.
       or (pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 1)   # PA_7_4 = 1 indica que fue agredido por alguién cercano con quien vive.
       or ((pd.notna(row.get("PB_2")) and row["PB_2"] == 1) and  # PB_2 = 1 especifica que fue agredido por algún adulto cercano.
       (pd.notna(row.get("PB_2_4")) and row["PB_2_4"] == 1))     # PB_2_4 = 1 indica que vive con el adulto cercano que lo agredió.
    ): 
        return "Persona cercana"

    if (pd.notna(row.get("PC_5")) and row["PC_5"] == 1):         # PC_5 = 1 especifica que fue agredido por una pareja romántica
        return "Pareja romántica"

    if (
       (pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 2)      # PA_6_4 = 2 indica que fue agredido por alguien cercano, pero con quien no vive.
       or (pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 2)   # PA_7_4 = 2 Esta alternativa indica que fue agredido por alguien cercano, pero con quien no vive.
       or ((pd.notna(row.get("PB_2")) and row["PB_2"] == 1) and  # PB_2 = 1 Esta pregunta especifica que fue agredido por algún adulto cercano.
       (pd.notna(row.get("PB_2_4")) and row["PB_2_4"] == 2))     # PB_2_4 = 2 Esta alternativa indica que no vive con el adulto cercano que lo agredió. 
    ): 
        return "Conocido"

    if (
       (pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 3)      # PA_6_4 = 3 indica que fue agredido por un extraño.
       or (pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 3)   # PA_7_4 = 3 esta alternativa indica que fue agredido por un extraño.
    ): 
        return "Extraño"

    return pd.NA

df_inicial = df_inicial.copy()
df_inicial["Tipo_agresor"] = df_inicial.apply(clasificar_agresor, axis=1)

conteos = (df_inicial.groupby(["Año", "Tipo_agresor"], observed=True).size().reset_index(name="Casos"))

# Calcular proporciones dentro de cada año
conteos["Porcentaje"] = (conteos.groupby("Año")["Casos"].transform(lambda x: x / x.sum() * 100))
proporciones = conteos.copy()

colores = ['#f8e590', '#c5e2e2','#d1b8a7',"#f2c2b6", '#d3d37e']
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for i, año in enumerate(sorted(df_inicial["Año"].unique())):
    datos = proporciones[proporciones["Año"] == año]
    axes[i].pie(datos["Porcentaje"], labels=datos["Tipo_agresor"], 
                colors=colores, autopct='%1.1f%%', startangle=90, wedgeprops={'width': 0.5, 'edgecolor': 'white'}, pctdistance=0.75)
    axes[i].set_title(f"\nDistribución de tipos de agresores\n ({año})", fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Creamos funciones para cada clasificasión
def pares_genero(row):
    if ((pd.notna(row.get("PA_5_4")) and row["PA_5_4"] == 3) 
        or (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 3)
        or (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 3)   # Estas alternativas especifican que sea adolescente o niño.
       ):
        return "Hombres"

    if ((pd.notna(row.get("PA_5_4")) and row["PA_5_4"] == 4)
        or (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 4)
        or (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 4)   # Estas alternativas especifican que sea adolescente o niña.
       ):
        return "Mujeres"

    return pd.NA


def persona_cercana_genero(row):
    if (((pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 1) and      # PA_6_4 = 1 indica que fue agredido por alguien cercano con quien vive.
         (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 1))         # PA_6_5 = 1 especifica que fue un adulto hombre.
        or ((pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 1) and   # PA_7_4 = 1 indica que fue agredido por alguién cercano con quien vive.
            (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 1))      # PA_7_5 = 1 espeficica que fue un adulto hombre.
       ):
        return "Hombres"

    if (((pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 1) and      # PA_6_4 = 1 indica que fue agredido por alguien cercano con quien vive.
         (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 2))         # PA_6_5 = 2 especifica que fue una adulta mujer.
        or ((pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 1) and   # PA_7_4 = 1 indica que fue agredido por alguién cercano con quien vive.
            (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 2))      # PA_7_5 = 2 espeficica que fue una adulta mujer.
       ):
        return "Mujeres"

    return pd.NA


def pareja_romantica_genero(row):
    if ((pd.notna(row.get("PC_5")) and row["PC_5"] == 1)
        and (pd.notna(row.get("PC_5_3")) and row["PC_5_3"] in [1, 3])  # Se especifica que fue agredido por una pareja romántica (Masculino).
       ):
        return "Hombres"

    if ((pd.notna(row.get("PC_5")) and row["PC_5"] == 1)
        and (pd.notna(row.get("PC_5_3")) and row["PC_5_3"] in [2, 4])  # Se especifica que fue agredido por una pareja romántica (Femenina).
       ):
        return "Mujeres"

    return pd.NA


def conocido_genero(row):
    if (((pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 2) and     # PA_6_4 = 2 indica que fue agredido por alguien cercano, pero con quien no vive.
         (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 1))        # PA_6_5 = 1 especifica que fue un adulto hombre.
        or ((pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 2) and  # PA_7_4 = 2 indica que fue agredido por alguien cercano, pero con quien no vive.
            (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 1))     # PA_7_5 = 1 especifica que fue un adulto hombre.
       ): 
        return "Hombres"

    if (((pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 2) and     # PA_6_4 = 2 indica que fue agredido por alguien cercano, pero con quien no vive.
         (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 2))        # PA_6_5 = 2 especifica que fue una adulta mujer.
        or ((pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 2) and  # PA_7_4 = 2 indica que fue agredido por alguien cercano, pero con quien no vive.
            (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 2))     # PA_7_5 = 2 especifica que fue una adulta mujer.
       ):
        return "Mujeres"

    return pd.NA


def extraño_genero(row):
    if (((pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 3) and     # PA_6_4 = 3 indica que fue agredido por un extraño.
         (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 1))        # PA_6_5 = 1 especifica que fue un adulto hombre.
        or ((pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 3) and  # PA_7_4 = 3 indica que fue agredido por un extraño.
            (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 1))     # PA_7_5 = 1 especifica que fue un adulto hombre.
       ):
        return "Hombres"

    if (((pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 3) and     # PA_6_4 = 3 indica que fue agredido por un extraño.
         (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 2))        # PA_6_5 = 2 especifica que fue una adulta mujer.
        or ((pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 3) and  # PA_7_4 = 3 indica que fue agredido por un extraño.
            (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 2))     # PA_7_5 = 2 especifica que fue una adulta mujer.
       ):
        return "Mujeres"

    return pd.NA


# Creamos una función para visualizar los gráficos
def graficar_genero(df, funcion_genero, colores, titulo):
    df = df.copy()
    df["Tipo_agresor"] = df.apply(funcion_genero, axis=1)

    conteos = (df.groupby(["Año", "Tipo_agresor"], observed=True).size().reset_index(name="Casos"))
    conteos["Porcentaje"] = conteos.groupby("Año")["Casos"].transform(lambda x: x / x.sum() * 100)

    fig, axes = plt.subplots(1, 2, figsize=(10, 4))
    for i, año in enumerate(sorted(df["Año"].unique())):
        datos = conteos[conteos["Año"] == año]
        axes[i].pie(datos["Porcentaje"], labels=datos["Tipo_agresor"],
                    colors=colores, autopct='%1.1f%%', startangle=90)
        axes[i].set_title(f"{titulo}\n ({año})", fontsize=10, fontweight='bold')

    plt.tight_layout()
    plt.show()

# Llamamos a la función para crear cada gráfico 
graficar_genero(df_inicial, pares_genero, ["#f2c2b6", "#ce8978"],
                "Distribución de género de los pares")
print(f'\n')
graficar_genero(df_inicial, persona_cercana_genero, ["#d3d37e", "#b5b53b"],
                "Distribución de género de las personas cercanas")
print(f'\n')
graficar_genero(df_inicial, pareja_romantica_genero, ["#d1b8a7", "#b08569"],
                "Distribución de género de las parejas románticas")
print(f'\n')
graficar_genero(df_inicial, conocido_genero, ["#f8e590", "#dfc349"],
                "Distribución de género de los conocidos")
print(f'\n')
graficar_genero(df_inicial, extraño_genero, ["#c5e2e2", "#88cece"],
                "Distribución de género de los extraños")
```

**Respuesta:** Al analizar los gráficos relacionados con la violencia física, se evidencia que los principales agresores son los mismos pares, representando más del 60 % en los dos años estudiados. Al entrar en más detalle, dentro de esta misma categoría, la mayoría de los agresores son hombres, correspondiendo a más del 70 % en los gráficos que distinguen por género de los atacantes.

La segunda categoría más frecuente son las personas cercanas, quienes son adultos que conviven con las víctimas encuestadas, representando aproximadamente un 15 % en los gráficos. A diferencia del caso anterior, al diferenciar a estos agresores por género, se observa que las mujeres constituyen la mayoría, superando el 60 % en ambos años.

Las siguientes categorías son considerablemente menos frecuentes: “Conocido”, “Extraño” y “Pareja romántica”, representando aproximadamente un 10 %, 7 % y 1,5 % respectivamente.

En el caso de los conocidos, se distingue que la mayoría de los agresores son hombres, con un 65 % en el gráfico del 2017. Sin embargo, al analizar el año 2023, se aprecia una disminución de este porcentaje, pasando los hombres a representar un 54,2 %, dejando muy equitativa la cantidad de agresiones provocadas por cada género.

Al referirse a los extraños, la diferencia entre la cantidad de agresores masculinos y femeninos es notable, siendo la mayoría hombres, representando aproximadamente un 80 % de las agresiones.

Finalmente, al analizar las parejas románticas, es interesante observar que la cantidad de agresiones por género es muy similar, prácticamente un 50 % para cada género.

### ***- Si comparamos ambas encuestas, realizadas con 6 años de diferencia, ¿se observa un aumento o una disminución en la cantidad y en las características de las vulneraciones reportadas?***

```python
df = pd.read_csv('Data-Encuestas.csv')

# Columnas de tipos de violencia
columnas_violencia_verbal = ['PB_1', 'PC_3']
columnas_violencia_fisica = ["PA_5", "PA_6", "PA_7", "PB_2", "PC_1", "PC_2", "PC_5"]
columnas_violencia_sexual = ["PD_3", "PD_4", "PD_5", "PD_6", "PD_7"]

año1 = 2017
año2 = 2023

# Función para calcular los porcentajes del tipo de violencia
def calcular_porcentajes(df, año, columnas, tipo):
    df_año = df[df['Año'] == año]
    # Se calcula la media de todos los valores '1' dentro de las columnas del tipo
    valores = []
    for col in columnas:
        if col in df_año.columns:
            valores.append((df_año[col] == 1).mean())
    if len(valores) > 0:
        return {
            'Año': año,
            'Tipo_de_Violencia': tipo,
            'Porcentaje': sum(valores) / len(valores) * 100
        }
    else:
        return None
resultados = []

for tipo, columnas in [('Verbal', columnas_violencia_verbal),
                       ('Física', columnas_violencia_fisica),
                       ('Sexual', columnas_violencia_sexual)]:
    for año in [año1, año2]:
        res = calcular_porcentajes(df, año, columnas, tipo)
        if res:
            resultados.append(res)

df_comparacion = pd.DataFrame(resultados)


# Gráfico
palette = {
    2017: '#dfdeea',
    2023: '#b27193'
}
plt.figure(figsize=(10, 6))
sns.barplot(
    data=df_comparacion,
    x='Tipo_de_Violencia',
    y='Porcentaje',
    hue='Año',
    palette= palette
)
plt.title('\nComparación de tipos de violencia infantil reportada\n (2017 vs 2023)\n', fontsize=18, fontweight='bold')
plt.xlabel('\nTipo de violencia\n', fontsize=12, color='grey')
plt.ylabel('\nPorcentaje de estudiantes que reportan\n haber sufrido violencia\n', fontsize=12, color='grey')
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.legend(title=' Año de encuesta ')
plt.tight_layout()
plt.show()


df_comparacion
```

### ***- ¿Existe una relación entre haber experimentado situaciones de violencia (física, sexual o verbal) y el bienestar emocional percibido por los estudiantes?***

```python
df_bienestar = df_inicial.copy()

# Columnas por tipo de violencia y bienestar
columnas_violencia_verbal = ['PB_1', 'PC_3']
columnas_violencia_fisica = ['PA_5', 'PA_6', 'PB_2', 'PC_1', 'PC_2', 'PC_5']
columnas_violencia_sexual = ['PD_3', 'PD_4', 'PD_5', 'PD_6', 'PD_7']
columnas_bienestar = ['PG_1', 'PG_2', 'PG_3', 'PG_4', 'PG_6', 'PG_8', 'PG_9', 'PG_10']

# Crear indicadores de violencia
for tipo, columnas in {
    'Verbal': columnas_violencia_verbal,
    'Física': columnas_violencia_fisica,
    'Sexual': columnas_violencia_sexual
}.items():
    df_bienestar[tipo] = df_bienestar[columnas].apply(lambda row: 1 if (row == 1).any() else 0, axis=1)

df_bienestar['Sufrió_Alguna_Violencia'] = df_bienestar[['Física', 'Verbal', 'Sexual']].max(axis=1)

# Promedio de las respuestas PG_1 a PG_10 (entre más alto → mayor bienestar)
df_bienestar['Indice_Bienestar'] = df_bienestar[columnas_bienestar].mean(axis=1)

color = {0: '#f3dcb3', 1: '#edbdbc'}

# GRÁFICA BIENESTAR GENERAL SEGÚN EXPERIENCIA DE VIOLENCIA
plt.figure(figsize=(10, 6))
ax = sns.boxplot(
    data=df_bienestar,
    x='Sufrió_Alguna_Violencia',
    y='Indice_Bienestar',
    hue='Sufrió_Alguna_Violencia',
    palette=color,
    width=0.7,
    dodge=False
)

# Corregir leyenda: ahora sí funciona
handles, _ = ax.get_legend_handles_labels()
ax.legend(handles, ['No', 'Sí'], title='¿Sufrió violencia?', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.title('\nBienestar emocional promedio según experiencia de violencia\n', fontsize=18, fontweight='bold')
plt.xlabel('\n¿Sufrió alguna violencia?\n', fontsize=12, color='grey')
plt.ylabel('Índice de bienestar emocional\n (promedio PG)\n', fontsize=12, color='grey')
plt.xticks([0, 1], ['No', 'Sí'])
plt.tight_layout()
plt.show()

print('\n')

# GRÁFICA BIENESTAR POR TIPO DE VIOLENCIA
df_melt = df_bienestar.melt(
    id_vars='Indice_Bienestar',
    value_vars=['Física', 'Verbal', 'Sexual'],
    var_name='Tipo_de_Violencia',
    value_name='Sufrió'
)

plt.figure(figsize=(10, 6))
ax = sns.boxplot(
    data=df_melt,
    x='Tipo_de_Violencia',
    y='Indice_Bienestar',
    hue='Sufrió',
    palette=color
)

ax.legend_.remove()
plt.title('\nComparación del bienestar emocional según tipo de violencia\n', fontsize=18, fontweight='bold')
plt.xlabel('\nTipo de violencia\n', fontsize=12, color='grey')
plt.ylabel('Índice de bienestar emocional\n (promedio PG)\n', fontsize=12, color='grey')
plt.tight_layout()
plt.show()
```

**Respuesta:** Existe una clara relación entre haber experimentado violencia y el bienestar emocional percibido. La mayoría de los estudiantes que reportan violencia (especialmente verbal y física) mantienen un bienestar emocional promedio. Sin embargo, el verdadero impacto de la violencia se observa en los extremos: sufrir cualquier tipo de violencia (verbal, física o sexual) está relacionado con la aparición de un grupo de estudiantes con un bienestar emocional críticamente bajo, algo que es mucho menos frecuente en quienes no reportan violencia.

```python
# Gráfico CORRELACIÓN ENTRE VIOLENCIAS Y BIENESTAR
# Seleccionamos las columnas relevantes
columnas_corr = [
    'Verbal',
    'Física',
    'Sexual',
    'Sufrió_Alguna_Violencia',
    'Indice_Bienestar'
]

# Calcular la matriz de correlación (Spearman)
corr_matrix = df_bienestar[columnas_corr].corr(method='spearman')

# Crear el mapa de calor
plt.figure(figsize=(8, 7))
sns.heatmap(
    corr_matrix,
    annot=True,                
    fmt=".2f",                 
    cmap='BrBG',            
    center=0,
    linewidths=0.5,            
    cbar_kws={'label': 'Correlación (Spearman)'}    
)

plt.title('\nCorrelación entre tipos de violencia y bienestar emocional\n', fontsize=18, fontweight='bold')
plt.xticks(rotation=45, ha='right')  
plt.yticks(rotation=0)               

plt.tight_layout()
plt.show()
```

### ***- ¿Cuál es la frecuencia con la que se cometen estos abusos a los niños?***

```python
frecuencias = {1: 'Nunca', 2: '1 vez', 3: '2-3 veces', 4: 'Al menos una vez\n al mes', 5: 'Al menos una vez\n a la semana', 6: 'Todos los días'}
tipos = {
    'Física': (['PA_5_2', 'PA_6_2', 'PA_7_2', 'PB_2_2', 'PC_1_2', 'PC_2_2', 'PC_5_2'], '#95785a'),
    'Verbal': (['PC_3_2', 'PB_1_2'], '#979b44'),
    'Sexual': (['PD_3_2', 'PD_4_2', 'PD_5_2', 'PD_6_2', 'PD_7_2'], '#dfdab0')
}

fig, axes = plt.subplots(3, 1, figsize=(10, 15))
for ax, (tipo, (columnas, color)) in zip(axes, tipos.items()):
    df = (
        df_fisica if tipo == 'Física' else
        df_verbal if tipo == 'Verbal' else
        df_sexual
    )

    # Contamos las frecuencias ignorando los valores 0 y 1
    conteo = pd.Series(dtype=int)
    for col in columnas:
        conteo = conteo.add(df[col][~df[col].isin([0, 1])].value_counts(), fill_value=0)

    # Reemplazamos los valores por las etiquetas
    conteo.index = [frecuencias[i] for i in conteo.index.astype(int)]
    conteo = conteo.sort_index()
    
    conteo.plot(kind='barh', color=color, ax=ax)
    ax.set_title(f'\nFrecuencia Violencia {tipo} últimos 12 meses\n', fontsize=18, fontweight='bold')
    ax.set_xlabel('\nCantidad de incidentes\n', fontsize=12, color='grey')
    ax.set_ylabel('Frecuencia\n', fontsize=12, color='grey')

plt.tight_layout()
plt.subplots_adjust(hspace=0.6)
plt.show()
```

**Respuesta:** Los registros indican que, en el caso de la violencia física, la mayoría de los niños y adolescentes que la han sufrido reportan haberla vivido alrededor de 2 a 3 veces (aproximadamente 11.200 casos), mientras que una minoría indica que la experimenta todos los días (aproximadamente 900 casos). En cuanto a la violencia verbal, la mayoría también reporta haberla sufrido 2 a 3 veces (aproximadamente 7.200 casos), y un grupo más pequeño indica que la vive a diario (unos 1.000 casos). Respecto a la violencia sexual, la mayoría de los registros muestran que los niños y adolescentes la han experimentado una sola vez (aproximadamente 1.200 casos), aunque también existen registros de quienes la sufren todos los días (alrededor de 100 casos).

### ***- De los casos de violencia sexual, ¿cuáles han sido reportados?***

```python
filtro_casos = {"PD_5": "PD_5_4", "PD_6": "PD_6_5", "PD_7": "PD_7_6"}

# Creamos un diccionario para registrar las cuentas de los casos
resultados = {}

for pregunta, subpregunta in filtro_casos.items():
    casos_sufridos = sexual[sexual[pregunta] == 1] # Filtramos por la primera pregunta
    n_sufridos = len(casos_sufridos)
    
    # De quienes respondieron "sí" en la primera pregunta, se filtran los que también respondieron "sí" en la segunda
    n_reportados = len(casos_sufridos[casos_sufridos[subpregunta] == 1])
    
    resultados[pregunta] = {
        "Casos sufridos": n_sufridos,
        "Casos reportados": n_reportados
    }

resultados = pd.DataFrame(resultados).T.reset_index()
resultados.rename(columns={"index": "Pregunta"}, inplace=True)

resultados["No reportados"] = resultados["Casos sufridos"] - resultados["Casos reportados"]
x = range(len(resultados))

# Graficamos
plt.figure(figsize=(8, 6))

plt.bar(x, resultados["Casos reportados"], width=0.5, color='#b2cca8', label="Casos reportados")
plt.bar(x, resultados["No reportados"], bottom=resultados["Casos reportados"], width=0.5, color='#cdd2e6', label="Sufridos NO reportados")

plt.title("\nComparación entre casos sufridos y reportados por tipo de violencia sexual\n", fontsize=18, fontweight='bold')
plt.xlabel("\nPregunta\n", fontsize=12, color='grey')
plt.ylabel("\nCantidad de casos\n", fontsize=12, color='grey')
plt.xticks(ticks=x, labels=resultados["Pregunta"])
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()

resultados
```

**Respuesta:** Se puede observar que, de los últimos casos de violencia sexual vividos por los niños, menos de la mitad ha sido reportada en todas las situaciones planteadas

```python
df = df_inicial.copy()
tipos = {
    "Física": ["PA_5", "PA_6", "PA_7", "PB_2", "PC_1", "PC_2", "PC_5"],
    "Verbal": ["PB_1", "PC_3"],
    "Sexual": ["PD_3", "PD_4", "PD_5", "PD_6", "PD_7"]
}

df_rm = df[df["Region_colegio"] == "Metropolitana"].copy()

# Normalizamos los nombres de comunas
def normalizar_texto(texto):
    if isinstance(texto, str):
        texto = texto.upper()
        texto = undt.normalize("NFKD", texto).encode("ascii", "ignore").decode("utf-8")
        return texto.strip()
    return texto

df_rm["Comuna_colegio"] = df_rm["Comuna_colegio"].apply(normalizar_texto)

# Creamos un Df que contenga los casos de violencia por tipo y su comuna
df_lista = []

for tipo, columnas in tipos.items():
    df_tmp = df_rm.copy()
    df_tmp['Casos'] = df_tmp[columnas].sum(axis=1)
    df_grouped = df_tmp.groupby('Comuna_colegio')['Casos'].sum().reset_index()
    df_grouped['Tipo'] = tipo
    df_lista.append(df_grouped)

df_burbujas = pd.concat(df_lista, ignore_index=True)


# Graficamos
plt.figure(figsize=(6, 35))
sns.scatterplot(
    data=df_burbujas,
    x="Tipo",
    y="Comuna_colegio",
    size="Casos",
    hue="Tipo",
    sizes=(500, 3000),
    alpha=0.6,
    palette="PuBu",
    legend=False
)

# Agregamos las etiquetas de las cantidad dentro de cada burbuja
for i in range(df_burbujas.shape[0]):
    x = df_burbujas['Tipo'].iloc[i]
    y = df_burbujas['Comuna_colegio'].iloc[i]
    value = df_burbujas['Casos'].iloc[i]
    plt.text(x, y, str(value), color='black', ha='center', va='center', fontsize=8)

plt.title("Casos de violencia por comuna y tipo\n (Región Metropolitana)\n", fontsize=18, fontweight="bold")
plt.xlabel("\nTipo de violencia\n", fontsize=14, color='grey')
plt.ylabel("\nComuna\n", fontsize=14, color='grey')
plt.margins(x=0.02, y=0.02)
plt.xlim(-0.3, 2.3) 
plt.tight_layout()
plt.show()
```
