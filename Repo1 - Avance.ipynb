# **Introducción**

En Chile y el mundo los niños, niñas y adolescentes son fundamentales para el desarrollo social, económico y cultural. Sin embargo, no es común hablar sobre el maltrato infantil, ya que muchas veces se percibe como un tema tabú o como algo perteneciente a problema del pasado. No obstante, en la actualidad continúan registrándose casos de abuso y violencia, tanto fuera como dentro de sus hogares.

Las consecuencias de estos actos tienen un impacto profundo en el desarrollo de los menores, afectando diferentes ámbitos como su salud física y mental, el rendimiento académico y las relaciones sociales. Estos efectos pueden generar, a largo plazo, dificultades emocionales, conductuales y sociales que repercuten en su bienestar y en su integración en la sociedad.

Por ello, es crucial visibilizar estos actos contra la niñez y abordar esta problemática de manera integral. Reconocer y analizar la magnitud de la violencia infantil permite diseñar estrategias de prevención, protección y apoyo, contribuyendo a garantizar un entorno seguro y propicio para el desarrollo de todos los niños y adolescentes.

# **Contexto y Motivación**

La niñez representa una etapa fundamental en el desarrollo humano, ya que marca diversos
procesos físicos, cognitivos y emocionales. Este período es sumamente importante, pues
durante él se desarrollan habilidades, valores y competencias que sientan las bases para la
vida adulta. La protección de los derechos infantiles y el establecimiento de relaciones
afectivas seguras y consistentes son esenciales, ya que los vínculos saludables, basados en
el cariño y en límites claros, favorecen la confianza, la autoestima y la capacidad de
relacionarse de manera positiva con los demás. Estos pilares permiten al niño desarrollar
seguridad emocional y social, creando una base sólida para enfrentar los desafíos de la vida
adulta.

Actualmente, en Chile existen 4.493.995 niños, niñas y adolescentes, lo que representa
aproximadamente el 24,3% de la población total del país. Este grupo de la población posee
derechos fundamentales que buscan garantizar su desarrollo integral, incluyendo el derecho
a la vida, educación, salud, libertad de expresión y a la protección frente a toda forma de
violencia, entre otros. No obstante, en la práctica, estos derechos no siempre se cumplen
plenamente, lo que puede generar impactos negativos en su bienestar físico, emocional y
social, limitando sus oportunidades de crecimiento y desarrollo pleno.

La violencia infantil constituye un problema tanto social como de salud pública, ya que genera
consecuencias profundas en el desarrollo emocional, social y educativo de quienes la sufren,
tanto a corto como a largo plazo. Los niños y adolescentes expuestos a violencia pueden
presentar dificultades de aprendizaje, problemas de conducta, afectación en sus relaciones
interpersonales y efectos negativos duraderos en su bienestar psicológico.

Es por esta razón que nuestra motivación se centra en comprender y evidenciar la existencia
de violencia hacia niños, niñas y adolescentes en Chile. Queremos destacar cómo los
distintos tipos de violencia afectan el bienestar emocional de quienes los sufren, permitiendo
identificar diferentes patrones, tendencias y cambios que surgen a lo largo del tiempo. Por
ello, nuestro público objetivo incluye diversos órganos gubernamentales, centros de salud,
establecimientos educacionales, familias y círculos cercanos, fuerzas de seguridad y
entidades dedicadas a la protección infantil, entre otros centros de interés. El objetivo es
generar conciencia sobre estas vulneraciones, con el fin de promover políticas, intervenciones
y acciones concretas que contribuyan a erradicar la violencia, garantizando un cuidado digno
y un desarrollo seguro para todos los niños, niñas y adolescentes de Chile

### ***¿Qué tipos de agresión (física, emocional, sexual) son los más frecuentes expuestos en los datos?***

### - **Fisicas**

```python
col = ["PA_5", "PA_6", "PA_7", "PB_2", "PC_1", "PC_2", "PC_5"]
fisica = df_fisica[df_fisica[col].eq(1).any(axis=1)]
```

### - **Verbal**

```python
col = ['PB_1', 'PC_3']
verbal = df_verbal[df_verbal[col].eq(1).any(axis=1)]
```

### - **Sexual**

```python
col = ["PD_3", "PD_4", "PD_5", "PD_6", "PD_7"]
sexual = df_sexual[df_sexual[col].eq(1).any(axis=1)]
```

```python
# Contamos los casos según el tipo de violencia por cada año 
conteo_F = fisica["Año"].value_counts().sort_index()
conteo_V = verbal["Año"].value_counts().sort_index()
conteo_S = sexual["Año"].value_counts().sort_index()

df_cant = pd.DataFrame({"Física": conteo_F, "Verbal": conteo_V, "Sexual": conteo_S}).fillna(0).astype(int)
df_cant
```

```python
# Graficamos
df_cant.plot(kind='bar', color=['green', 'brown', 'grey']) 

plt.title("\n Tipos de agresión por año\n",fontsize=18, fontweight='bold')
plt.xlabel("\nAño", fontsize= 12, color='grey')
plt.ylabel("Cantidad de personas\n", fontsize=12, color='grey')

plt.xticks(rotation=0)
plt.legend(title="Tipo de agresión")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
# Graficamos
titulos = ['\nViolencia Física', '\nViolencia Verbal', '\nViolencia Sexual']
dataframes = [fisica, verbal, sexual]

fig, axes = plt.subplots(1, 3, figsize=(14, 10))
for ax, df, titulo in zip(axes, dataframes, titulos):
    conteo = df['Sexo'].value_counts()
    colores = [
        'pink' if sexo == 'Mujer' else 'blue'
        for sexo in conteo.index
    ]
    ax.pie(
        conteo,
        labels = conteo.index,
        autopct = '%1.1f%%',
        startangle = 90,
        colors = colores,
        wedgeprops = {'edgecolor': 'white'}
    )
    ax.set_title(titulo, fontsize=18, fontweight='bold')

plt.tight_layout()
plt.subplots_adjust(wspace=0.3)
plt.show()
```

```python
# Graficamos
violencias = ['Física', 'Verbal', 'Sexual']
dfs = [fisica, verbal, sexual]
años = [2017, 2023]

fig, axes = plt.subplots(3, 2, figsize=(10, 12))
for i, (nombre, df) in enumerate(zip(violencias, dfs)):
    for j, año in enumerate(años):
        df_año = df[df['Año'] == año]
        conteo = df_año['Sexo'].value_counts()

        colores = [
            'pink' if sexo == 'Mujer' else 'blue'
            for sexo in conteo.index
        ]

        axes[i, j].pie(
            conteo,
            labels = conteo.index,
            autopct = '%1.1f%%',
            startangle = 90,
            colors = colores,
            wedgeprops = {'edgecolor': 'white'}
        )

        axes[i, j].set_title(f'Violencia {nombre}\n {año}', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.subplots_adjust(hspace=0.3)
plt.show()
```

```python
frecuencias = {1: 'Nunca', 2: '1 vez', 3: '2-3 veces', 4: 'Al menos una vez\n al mes', 5: 'Al menos una vez\n a la semana', 6: 'Todos los días'}
tipos = {
    'Física': (['PA_5_2', 'PA_6_2', 'PA_7_2', 'PB_2_2', 'PC_1_2', 'PC_2_2', 'PC_5_2'], 'blue'),
    'Verbal': (['PC_3_2', 'PB_1_2'], 'green'),
    'Sexual': (['PD_3_2', 'PD_4_2', 'PD_5_2', 'PD_6_2', 'PD_7_2'], 'orange')
}

fig, axes = plt.subplots(3, 1, figsize=(10, 15))
for ax, (tipo, (columnas, color)) in zip(axes, tipos.items()):
    df = (
        df_fisica if tipo == 'Física' else
        df_verbal if tipo == 'Verbal' else
        df_sexual
    )

    # Contamos las frecuencias ignorando los valores 0 y 1
    conteo = pd.Series(dtype=int)
    for col in columnas:
        conteo = conteo.add(df[col][~df[col].isin([0, 1])].value_counts(), fill_value=0)

    # Reemplazamos los valores por las etiquetas
    conteo.index = [frecuencias[i] for i in conteo.index.astype(int)]
    conteo = conteo.sort_index()
    
    conteo.plot(kind='barh', color=color, edgecolor='black', ax=ax)
    ax.set_title(f'\nFrecuencia Violencia {tipo} últimos 12 meses\n', fontsize=18, fontweight='bold')
    ax.set_xlabel('\nCantidad de incidentes\n', fontsize=12, color='grey')
    ax.set_ylabel('Frecuencia\n', fontsize=12, color='grey')

plt.tight_layout()
plt.subplots_adjust(hspace=0.6)
plt.show()
```

### ***¿Qué proporción de niños, niñas y adolescentes ha experimentado situaciones de violencia sexual en sus distintas manifestaciones?***

```python
colum = ['PD_3', 'PD_4', 'PD_5', 'PD_6', 'PD_7']
colores = ['#4A90E2', '#E94E77']  # Hombre, Mujer

conteo = {}
for pd_col in colum:
    conteo[pd_col] = df_sexual.groupby('Sexo')[pd_col].apply(lambda x: (x == 1).sum())

conteo_df = pd.DataFrame(conteo).T
conteo_df.index = [
    'Exposición forzada o sorpresiva\n de partes íntimas', 
    'Forzar u obligar a realizar\n actos de carácter sexual', 
    'Toque forzado de partes íntimas\n por adulto extraño', 
    'Toque forzado de partes íntimas\n por adulto conocido', 
    'Relaciones sexuales\n forzadas o tentaviva'
]

conteo_df.plot(kind='bar', figsize=(10,6), color=colores)
plt.title('\nCantidad de experiencias de violencia sexual por tipo y sexo\n', fontsize=18, fontweight='bold')
plt.xlabel('\nTipos de Violencias Sexuales\n', fontsize=12, color='grey')
plt.ylabel('Cantidad de niños\n', fontsize=12, color='grey')

plt.legend(title='Sexo')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```
