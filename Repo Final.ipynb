# **Introducción**

En Chile y el mundo los niños, niñas y adolescentes son fundamentales para el desarrollo social, económico y cultural. Sin embargo, no es común hablar sobre el maltrato infantil, ya que muchas veces se percibe como un tema tabú o como algo perteneciente a problema del pasado. No obstante, en la actualidad continúan registrándose casos de abuso y violencia, tanto fuera como dentro de sus hogares.

Las consecuencias de estos actos tienen un impacto profundo en el desarrollo de los menores, afectando diferentes ámbitos como su salud física y mental, el rendimiento académico y las relaciones sociales. Estos efectos pueden generar, a largo plazo, dificultades emocionales, conductuales y sociales que repercuten en su bienestar y en su integración en la sociedad.

Por ello, es crucial visibilizar estos actos contra la niñez y abordar esta problemática de manera integral. Reconocer y analizar la magnitud de la violencia infantil permite diseñar estrategias de prevención, protección y apoyo, contribuyendo a garantizar un entorno seguro y propicio para el desarrollo de todos los niños y adolescentes.

# **Contexto y Motivación**

La niñez representa una etapa fundamental en el desarrollo humano, ya que marca diversos
procesos físicos, cognitivos y emocionales. Este período es sumamente importante, pues
durante él se desarrollan habilidades, valores y competencias que sientan las bases para la
vida adulta. La protección de los derechos infantiles y el establecimiento de relaciones
afectivas seguras y consistentes son esenciales, ya que los vínculos saludables, basados en
el cariño y en límites claros, favorecen la confianza, la autoestima y la capacidad de
relacionarse de manera positiva con los demás. Estos pilares permiten al niño desarrollar
seguridad emocional y social, creando una base sólida para enfrentar los desafíos de la vida
adulta.

Actualmente, en Chile existen 4.493.995 niños, niñas y adolescentes, lo que representa
aproximadamente el 24,3% de la población total del país. Este grupo de la población posee
derechos fundamentales que buscan garantizar su desarrollo integral, incluyendo el derecho
a la vida, educación, salud, libertad de expresión y a la protección frente a toda forma de
violencia, entre otros. No obstante, en la práctica, estos derechos no siempre se cumplen
plenamente, lo que puede generar impactos negativos en su bienestar físico, emocional y
social, limitando sus oportunidades de crecimiento y desarrollo pleno.

La violencia infantil constituye un problema tanto social como de salud pública, ya que genera
consecuencias profundas en el desarrollo emocional, social y educativo de quienes la sufren,
tanto a corto como a largo plazo. Los niños y adolescentes expuestos a violencia pueden
presentar dificultades de aprendizaje, problemas de conducta, afectación en sus relaciones
interpersonales y efectos negativos duraderos en su bienestar psicológico.

Es por esta razón que nuestra motivación se centra en comprender y evidenciar la existencia
de violencia hacia niños, niñas y adolescentes en Chile. Queremos destacar cómo los
distintos tipos de violencia afectan el bienestar emocional de quienes los sufren, permitiendo
identificar diferentes patrones, tendencias y cambios que surgen a lo largo del tiempo. Por
ello, nuestro público objetivo incluye diversos órganos gubernamentales, centros de salud,
establecimientos educacionales, familias y círculos cercanos, fuerzas de seguridad y
entidades dedicadas a la protección infantil, entre otros centros de interés. El objetivo es
generar conciencia sobre estas vulneraciones, con el fin de promover políticas, intervenciones
y acciones concretas que contribuyan a erradicar la violencia, garantizando un cuidado digno
y un desarrollo seguro para todos los niños, niñas y adolescentes de Chile

### ***- ¿Cuál es la proporción de casos de violencia física atribuida a pares (otros niños, niñas o adolescentes), personas cercanas, conocidos y extraños dentro de los datos?***

```python
violencia_fisica = ["PA_5_4","PA_6_4","PA_6_5","PA_7_4","PA_7_5","PB_2","PB_2_4","PC_1","PC_2","PC_5","PC_5_3"]
df_inicial[violencia_fisica] = df_inicial[violencia_fisica].replace(0, pd.NA)
df_inicial = df_inicial[df_inicial[violencia_fisica].notna().any(axis=1)]
```

```python
def clasificar_agresor(row):
    if (
       (pd.notna(row.get("PA_5_4")) and row["PA_5_4"] in [3,4])  # "PA_5_4 = [3,4] indica que sea adolescente, niño o niña.
       or (pd.notna(row.get("PC_1")) and row["PC_1"] == 1)       # PC_1 = 1 especifica que fue agredido por un niño, niña o adolescente.
       or (pd.notna(row.get("PC_2")) and row["PC_2"] == 1)       # PC_2 = 1 especifica que fue agredido por un grupo de niños, niñas o adolescentes.
    ): 
        return "Pares"

    if (
       (pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 1)      # PA_6_4 = 1 indica que fue agredido por alguien cercano con quien vive.
       or (pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 1)   # PA_7_4 = 1 indica que fue agredido por alguién cercano con quien vive.
       or ((pd.notna(row.get("PB_2")) and row["PB_2"] == 1) and  # PB_2 = 1 especifica que fue agredido por algún adulto cercano.
       (pd.notna(row.get("PB_2_4")) and row["PB_2_4"] == 1))     # PB_2_4 = 1 indica que vive con el adulto cercano que lo agredió.
    ): 
        return "Persona cercana"

    if (pd.notna(row.get("PC_5")) and row["PC_5"] == 1):         # PC_5 = 1 especifica que fue agredido por una pareja romántica
        return "Pareja romántica"

    if (
       (pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 2)      # PA_6_4 = 2 indica que fue agredido por alguien cercano, pero con quien no vive.
       or (pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 2)   # PA_7_4 = 2 Esta alternativa indica que fue agredido por alguien cercano, pero con quien no vive.
       or ((pd.notna(row.get("PB_2")) and row["PB_2"] == 1) and  # PB_2 = 1 Esta pregunta especifica que fue agredido por algún adulto cercano.
       (pd.notna(row.get("PB_2_4")) and row["PB_2_4"] == 2))     # PB_2_4 = 2 Esta alternativa indica que no vive con el adulto cercano que lo agredió. 
    ): 
        return "Conocido"

    if (
       (pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 3)      # PA_6_4 = 3 indica que fue agredido por un extraño.
       or (pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 3)   # PA_7_4 = 3 esta alternativa indica que fue agredido por un extraño.
    ): 
        return "Extraño"

    return pd.NA

df_inicial = df_inicial.copy()
df_inicial["Tipo_agresor"] = df_inicial.apply(clasificar_agresor, axis=1)

conteos = (df_inicial.groupby(["Año", "Tipo_agresor"], observed=True).size().reset_index(name="Casos"))

# Calcular proporciones dentro de cada año
conteos["Porcentaje"] = (conteos.groupby("Año")["Casos"].transform(lambda x: x / x.sum() * 100))
proporciones = conteos.copy()

colores = ['#f8e590', '#c5e2e2','#d1b8a7',"#f2c2b6", '#d3d37e']
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for i, año in enumerate(sorted(df_inicial["Año"].unique())):
    datos = proporciones[proporciones["Año"] == año]
    axes[i].pie(datos["Porcentaje"], labels=datos["Tipo_agresor"], 
                colors=colores, autopct='%1.1f%%', startangle=90, wedgeprops={'width': 0.5, 'edgecolor': 'white'}, pctdistance=0.75)
    axes[i].set_title(f"\nDistribución de tipos de agresores\n ({año})", fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Creamos funciones para cada clasificasión
def pares_genero(row):
    if ((pd.notna(row.get("PA_5_4")) and row["PA_5_4"] == 3) 
        or (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 3)
        or (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 3)   # Estas alternativas especifican que sea adolescente o niño.
       ):
        return "Hombres"

    if ((pd.notna(row.get("PA_5_4")) and row["PA_5_4"] == 4)
        or (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 4)
        or (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 4)   # Estas alternativas especifican que sea adolescente o niña.
       ):
        return "Mujeres"

    return pd.NA


def persona_cercana_genero(row):
    if (((pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 1) and      # PA_6_4 = 1 indica que fue agredido por alguien cercano con quien vive.
         (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 1))         # PA_6_5 = 1 especifica que fue un adulto hombre.
        or ((pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 1) and   # PA_7_4 = 1 indica que fue agredido por alguién cercano con quien vive.
            (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 1))      # PA_7_5 = 1 espeficica que fue un adulto hombre.
       ):
        return "Hombres"

    if (((pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 1) and      # PA_6_4 = 1 indica que fue agredido por alguien cercano con quien vive.
         (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 2))         # PA_6_5 = 2 especifica que fue una adulta mujer.
        or ((pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 1) and   # PA_7_4 = 1 indica que fue agredido por alguién cercano con quien vive.
            (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 2))      # PA_7_5 = 2 espeficica que fue una adulta mujer.
       ):
        return "Mujeres"

    return pd.NA


def pareja_romantica_genero(row):
    if ((pd.notna(row.get("PC_5")) and row["PC_5"] == 1)
        and (pd.notna(row.get("PC_5_3")) and row["PC_5_3"] in [1, 3])  # Se especifica que fue agredido por una pareja romántica (Masculino).
       ):
        return "Hombres"

    if ((pd.notna(row.get("PC_5")) and row["PC_5"] == 1)
        and (pd.notna(row.get("PC_5_3")) and row["PC_5_3"] in [2, 4])  # Se especifica que fue agredido por una pareja romántica (Femenina).
       ):
        return "Mujeres"

    return pd.NA


def conocido_genero(row):
    if (((pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 2) and     # PA_6_4 = 2 indica que fue agredido por alguien cercano, pero con quien no vive.
         (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 1))        # PA_6_5 = 1 especifica que fue un adulto hombre.
        or ((pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 2) and  # PA_7_4 = 2 indica que fue agredido por alguien cercano, pero con quien no vive.
            (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 1))     # PA_7_5 = 1 especifica que fue un adulto hombre.
       ): 
        return "Hombres"

    if (((pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 2) and     # PA_6_4 = 2 indica que fue agredido por alguien cercano, pero con quien no vive.
         (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 2))        # PA_6_5 = 2 especifica que fue una adulta mujer.
        or ((pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 2) and  # PA_7_4 = 2 indica que fue agredido por alguien cercano, pero con quien no vive.
            (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 2))     # PA_7_5 = 2 especifica que fue una adulta mujer.
       ):
        return "Mujeres"

    return pd.NA


def extraño_genero(row):
    if (((pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 3) and     # PA_6_4 = 3 indica que fue agredido por un extraño.
         (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 1))        # PA_6_5 = 1 especifica que fue un adulto hombre.
        or ((pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 3) and  # PA_7_4 = 3 indica que fue agredido por un extraño.
            (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 1))     # PA_7_5 = 1 especifica que fue un adulto hombre.
       ):
        return "Hombres"

    if (((pd.notna(row.get("PA_6_4")) and row["PA_6_4"] == 3) and     # PA_6_4 = 3 indica que fue agredido por un extraño.
         (pd.notna(row.get("PA_6_5")) and row["PA_6_5"] == 2))        # PA_6_5 = 2 especifica que fue una adulta mujer.
        or ((pd.notna(row.get("PA_7_4")) and row["PA_7_4"] == 3) and  # PA_7_4 = 3 indica que fue agredido por un extraño.
            (pd.notna(row.get("PA_7_5")) and row["PA_7_5"] == 2))     # PA_7_5 = 2 especifica que fue una adulta mujer.
       ):
        return "Mujeres"

    return pd.NA


# Creamos una función para visualizar los gráficos
def graficar_genero(df, funcion_genero, colores, titulo):
    df = df.copy()
    df["Tipo_agresor"] = df.apply(funcion_genero, axis=1)

    conteos = (df.groupby(["Año", "Tipo_agresor"], observed=True).size().reset_index(name="Casos"))
    conteos["Porcentaje"] = conteos.groupby("Año")["Casos"].transform(lambda x: x / x.sum() * 100)

    fig, axes = plt.subplots(1, 2, figsize=(10, 4))
    for i, año in enumerate(sorted(df["Año"].unique())):
        datos = conteos[conteos["Año"] == año]
        axes[i].pie(datos["Porcentaje"], labels=datos["Tipo_agresor"],
                    colors=colores, autopct='%1.1f%%', startangle=90)
        axes[i].set_title(f"{titulo}\n ({año})", fontsize=10, fontweight='bold')

    plt.tight_layout()
    plt.show()

# Llamamos a la función para crear cada gráfico 
graficar_genero(df_inicial, pares_genero, ["#f2c2b6", "#ce8978"],
                "Distribución de género de los pares")
print(f'\n')
graficar_genero(df_inicial, persona_cercana_genero, ["#d3d37e", "#b5b53b"],
                "Distribución de género de las personas cercanas")
print(f'\n')
graficar_genero(df_inicial, pareja_romantica_genero, ["#d1b8a7", "#b08569"],
                "Distribución de género de las parejas románticas")
print(f'\n')
graficar_genero(df_inicial, conocido_genero, ["#f8e590", "#dfc349"],
                "Distribución de género de los conocidos")
print(f'\n')
graficar_genero(df_inicial, extraño_genero, ["#c5e2e2", "#88cece"],
                "Distribución de género de los extraños")
```

**Respuesta:** Al analizar los gráficos relacionados con la violencia física, se evidencia que los principales agresores son los mismos pares, representando más del 60 % en los dos años estudiados. Al entrar en más detalle, dentro de esta misma categoría, la mayoría de los agresores son hombres, correspondiendo a más del 70 % en los gráficos que distinguen por género de los atacantes.

La segunda categoría más frecuente son las personas cercanas, quienes son adultos que conviven con las víctimas encuestadas, representando aproximadamente un 15 % en los gráficos. A diferencia del caso anterior, al diferenciar a estos agresores por género, se observa que las mujeres constituyen la mayoría, superando el 60 % en ambos años.

Las siguientes categorías son considerablemente menos frecuentes: “Conocido”, “Extraño” y “Pareja romántica”, representando aproximadamente un 10 %, 7 % y 1,5 % respectivamente.

```python
df = pd.read_csv('Data-Encuestas.csv')

# Columnas de tipos de violencia
columnas_violencia_verbal = ['PB_1', 'PC_3']
columnas_violencia_fisica = ["PA_5", "PA_6", "PA_7", "PB_2", "PC_1", "PC_2", "PC_5"]
columnas_violencia_sexual = ["PD_3", "PD_4", "PD_5", "PD_6", "PD_7"]

año1 = 2017
año2 = 2023

# Función para calcular los porcentajes del tipo de violencia
def calcular_porcentajes(df, año, columnas, tipo):
    df_año = df[df['Año'] == año]
    # Se calcula la media de todos los valores '1' dentro de las columnas del tipo
    valores = []
    for col in columnas:
        if col in df_año.columns:
            valores.append((df_año[col] == 1).mean())
    if len(valores) > 0:
        return {
            'Año': año,
            'Tipo_de_Violencia': tipo,
            'Porcentaje': sum(valores) / len(valores) * 100
        }
    else:
        return None
resultados = []

for tipo, columnas in [('Verbal', columnas_violencia_verbal),
                       ('Física', columnas_violencia_fisica),
                       ('Sexual', columnas_violencia_sexual)]:
    for año in [año1, año2]:
        res = calcular_porcentajes(df, año, columnas, tipo)
        if res:
            resultados.append(res)

df_comparacion = pd.DataFrame(resultados)


# Gráfico
palette = {
    2017: '#dfdeea',
    2023: '#b27193'
}
plt.figure(figsize=(10, 6))
sns.barplot(
    data=df_comparacion,
    x='Tipo_de_Violencia',
    y='Porcentaje',
    hue='Año',
    palette= palette
)
plt.title('\nComparación de tipos de violencia infantil reportada\n (2017 vs 2023)\n', fontsize=18, fontweight='bold')
plt.xlabel('\nTipo de violencia\n', fontsize=12, color='grey')
plt.ylabel('\nPorcentaje de estudiantes que reportan\n haber sufrido violencia\n', fontsize=12, color='grey')
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.legend(title=' Año de encuesta ')
plt.tight_layout()
plt.show()


df_comparacion
```

### ***- ¿Existe una relación entre haber experimentado situaciones de violencia (física, sexual o verbal) y el bienestar emocional percibido por los estudiantes?***

```python
df_bienestar = df_inicial.copy()

# Columnas por tipo de violencia y bienestar
columnas_violencia_verbal = ['PB_1', 'PC_3']
columnas_violencia_fisica = ['PA_5', 'PA_6', 'PB_2', 'PC_1', 'PC_2', 'PC_5']
columnas_violencia_sexual = ['PD_3', 'PD_4', 'PD_5', 'PD_6', 'PD_7']
columnas_bienestar = ['PG_1', 'PG_2', 'PG_3', 'PG_4', 'PG_6', 'PG_8', 'PG_9', 'PG_10']

# Crear indicadores de violencia
for tipo, columnas in {
    'Verbal': columnas_violencia_verbal,
    'Física': columnas_violencia_fisica,
    'Sexual': columnas_violencia_sexual
}.items():
    df_bienestar[tipo] = df_bienestar[columnas].apply(lambda row: 1 if (row == 1).any() else 0, axis=1)

df_bienestar['Sufrió_Alguna_Violencia'] = df_bienestar[['Física', 'Verbal', 'Sexual']].max(axis=1)

# Promedio de las respuestas PG_1 a PG_10 (entre más alto → mayor bienestar)
df_bienestar['Indice_Bienestar'] = df_bienestar[columnas_bienestar].mean(axis=1)

color = {0: '#f3dcb3', 1: '#edbdbc'}

# GRÁFICA BIENESTAR GENERAL SEGÚN EXPERIENCIA DE VIOLENCIA
plt.figure(figsize=(10, 6))
ax = sns.boxplot(
    data=df_bienestar,
    x='Sufrió_Alguna_Violencia',
    y='Indice_Bienestar',
    hue='Sufrió_Alguna_Violencia',
    palette=color,
    width=0.7,
    dodge=False
)

# Corregir leyenda: ahora sí funciona
handles, _ = ax.get_legend_handles_labels()
ax.legend(handles, ['No', 'Sí'], title='¿Sufrió violencia?', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.title('\nBienestar emocional promedio según experiencia de violencia\n', fontsize=18, fontweight='bold')
plt.xlabel('\n¿Sufrió alguna violencia?\n', fontsize=12, color='grey')
plt.ylabel('Índice de bienestar emocional\n (promedio PG)\n', fontsize=12, color='grey')
plt.xticks([0, 1], ['No', 'Sí'])
plt.tight_layout()
plt.show()

print('\n')

# GRÁFICA BIENESTAR POR TIPO DE VIOLENCIA
df_melt = df_bienestar.melt(
    id_vars='Indice_Bienestar',
    value_vars=['Física', 'Verbal', 'Sexual'],
    var_name='Tipo_de_Violencia',
    value_name='Sufrió'
)

plt.figure(figsize=(10, 6))
ax = sns.boxplot(
    data=df_melt,
    x='Tipo_de_Violencia',
    y='Indice_Bienestar',
    hue='Sufrió',
    palette=color
)

ax.legend_.remove()
plt.title('\nComparación del bienestar emocional según tipo de violencia\n', fontsize=18, fontweight='bold')
plt.xlabel('\nTipo de violencia\n', fontsize=12, color='grey')
plt.ylabel('Índice de bienestar emocional\n (promedio PG)\n', fontsize=12, color='grey')
plt.tight_layout()
plt.show()
```

**Respuesta:** Existe una clara relación entre haber experimentado violencia y el bienestar emocional percibido. La mayoría de los estudiantes que reportan violencia (especialmente verbal y física) mantienen un bienestar emocional promedio. Sin embargo, el verdadero impacto de la violencia se observa en los extremos: sufrir cualquier tipo de violencia (verbal, física o sexual) está relacionado con la aparición de un grupo de estudiantes con un bienestar emocional críticamente bajo, algo que es mucho menos frecuente en quienes no reportan violencia.

En este análisis, nuestro objetivo principal es evidenciar la magnitud y características de la violencia que afecta a niños, niñas y adolescentes en Chile, así como examinar su evolución en el tiempo. Estas encuestas permiten abordar nuestra pregunta central:

***¿Cómo se manifiestan los distintos tipos de violencia hacia niños, niñas y adolescentes en Chile, de qué manera afectan su bienestar emocional y cómo ha cambiado su prevalencia a lo largo del tiempo?***

- violencia verbal
- violencia sexual.

- 21.000 casos de violencia física

- 4.500 casos de violencia sexual

#### **3. Frecuencia y características de los casos**

- La frecuencia de cada tipo de violencia
- La distribución según sexo y edad

A partir de estas proporciones, se observa que no existe un patrón único ni consistente que indique que un sexo experimente sistemáticamente más violencia que el otro.

### ***2. ¿Existe una relación entre haber experimentado situaciones de violencia (física, sexual o verbal) y el bienestar emocional percibido por los  estudiantes?*** 

#### **- Metodología**
#### a) Construcción de variables binarias de violencia

Para evaluar la relación entre la experiencia de violencia y el bienestar emocional, se crearon variables  que identifican si cada estudiante había sufrido violencia física, verbal o sexual. Cada variable toma el valor de 1 cuando el estudiante reporta al menos una vulneración dentro de la categoría correspondiente y 0 en caso contrario.

A partir de estas tres variables, se definió una variable agregada que indica si el estudiante experimentó cualquier tipo de violencia:

$$
\text{Sufrió\_Alguna\_Violencia} = \max(\text{Física},\ \text{Verbal},\ \text{Sexual})
$$



Esta construcción permite distinguir entre estudiantes expuestos a diferentes tipos de vulneración y aquellos sin experiencias reportadas, y sirve como base para los análisis posteriores.


#### b) Cálculo del Índice de Bienestar Emocional

El bienestar emocional se operacionalizó mediante un índice compuesto calculado como el promedio de las respuestas a los ítems PG_1, PG_2, PG_3, PG_4, PG_6, PG_8, PG_9 y PG_10. El uso de un promedio simple implica que cada dimensión del bienestar contribuye de manera equivalente al puntaje final, lo que permite una interpretación directa: valores más altos reflejan un mayor nivel de bienestar subjetivo.


#### c) Visualizaciones utilizadas

Con el fin de explorar la relación entre las variables, se elaboraron distintas representaciones gráficas:

Un boxplot del índice de bienestar según si el estudiante sufrió o no alguna forma de violencia, con el objetivo de comparar las distribuciones entre ambos grupos.

Un boxplot del índice de bienestar desagregado por tipo de violencia (física, verbal y sexual), lo que permite identificar cuál de estas formas presenta una mayor asociación con niveles reducidos de bienestar.

Un mapa de calor basado en correlaciones de Spearman, que resume la intensidad y dirección de las asociaciones entre los tipos de violencia y el índice de bienestar emocional.

Estas visualizaciones complementan el análisis estadístico al permitir observar patrones de dispersión, valores atípicos y diferencias en las distribuciones.


#### d) Métrica estadística utilizada

La relación entre violencia y bienestar emocional se evaluó mediante el coeficiente de correlación de Spearman. Esta métrica fue seleccionada debido a su pertinencia para datos que no necesariamente presentan relaciones lineales, su adecuación para variables en escala ordinal y su robustez ante distribuciones asimétricas, lo que resulta apropiado para el análisis de fenómenos psicosociales como los estudiados en este proyecto.

#### **- Modelo Validación Estadística**

```python
# MODELO LOGÍSTICO
X = df_bienestar[['Indice_Bienestar']]
y = df_bienestar['Sufrió_Alguna_Violencia']

# Separar datos
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Entrenamiento
modelo = LogisticRegression()
modelo.fit(X_train, y_train)

# Predicciones
y_pred = modelo.predict(X_test)
y_prob = modelo.predict_proba(X_test)[:, 1]  # Probabilidad de clase positiva

# MÉTRICAS
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_prob)

print("\n\n=====  MÉTRICAS DEL MODELO =====")
print(f"Accuracy:  {acc:.3f}")
print(f"Precision: {prec:.3f}")
print(f"Recall:    {rec:.3f}")
print(f"F1-score:  {f1:.3f}")
print(f"AUC-ROC:   {auc:.3f}")

# MATRIZ DE CONFUSIÓN
plt.figure(figsize=(6,4))
ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred)).plot(cmap="YlGn")
plt.title("\n\nMatriz de Confusión — Modelo de Violencia vs Bienestar\n", fontsize=18, fontweight="bold")
plt.show()
```

Se entrenó un modelo de regresión logística utilizando el Índice de Bienestar como predictor y la variable “Sufrió_Alguna_Violencia” como variable objetivo. Los resultados muestran que, aunque el bienestar emocional es un predictor significativo, su capacidad de clasificar correctamente a los estudiantes es moderada.

El modelo presenta valores de precisión, recall y F1-score que reflejan que las personas con niveles muy bajos de bienestar son clasificadas correctamente con mayor facilidad, mientras que los casos intermedios resultan más difíciles de distinguir. El AUC-ROC se mantiene sobre 0.60, indicando que el bienestar posee poder discriminativo, aunque limitado, para identificar experiencias de violencia.

La matriz de confusión revela que el modelo identifica adecuadamente a una proporción importante de quienes han sufrido violencia (recall), pero presenta dificultades para clasificar a estudiantes sin violencia, debido a la alta prevalencia general del fenómeno en la muestra.

Los resultados del modelo indican que, aunque existe una asociación entre el bienestar emocional y la experiencia de violencia, esta relación es limitada para propósitos predictivos. El modelo logra un recall de 1.0, lo que significa que identifica correctamente a todos los estudiantes que efectivamente reportaron haber sufrido violencia, pero al mismo tiempo clasifica erróneamente a todos los estudiantes sin violencia como si sí la hubieran sufrido. Esto se refleja en una matriz de confusión donde no se registran verdaderos negativos y donde la precisión y el accuracy se mantienen en torno al 76%, coincidiendo con la alta prevalencia de violencia observada en la muestra. El valor AUC de 0.59 confirma que el índice de bienestar posee un poder discriminativo bajo y apenas superior al azar. En términos prácticos, estos resultados sugieren que los niveles de bienestar emocional están relacionados con la experiencia de violencia, especialmente en los extremos más bajos del bienestar; sin embargo, el bienestar por sí solo no constituye un predictor suficientemente robusto como para diferenciar adecuadamente entre estudiantes que han sufrido violencia y aquellos que no.

### ***3. Si comparamos ambas encuestas, realizadas con 6 años de diferencia, ¿se observa un aumento o una disminución en la cantidad y en las características de las vulneraciones reportadas?***

#### **- Metodología**
#### (a) Selección de variables
Para analizar los cambios en la prevalencia de vulneraciones, se definieron tres categorías principales de violencia, agrupando según la naturaleza de la vulneración reportada. La siguiente tabla resume la clasificación:

| Categoría        | Columnas relacionadas                    |
| ---------------- | ---------------------------------------- |
| Violencia verbal | PB_1, PC_3                               |
| Violencia física | PA_5, PA_6, PA_7, PB_2, PC_1, PC_2, PC_5 |
| Violencia sexual | PD_3, PD_4, PD_5, PD_6, PD_7             |

Estas categorías permiten comparar la magnitud y evolución de los distintos tipos de violencia incluidos en ambas encuestas (2017 y 2023).

#### (b) Cálculo de prevalencia
La prevalencia de violencia por categoría se estimó como el porcentaje de estudiantes que reportan haber sufrido al menos una vulneración dentro de cada conjunto de ítems. Para ello, se calculó la proporción de respuestas afirmativas en cada categoría, aplicando la siguiente expresión:

$$ Porcentaje = \frac{\sum \mathbf{1}(\text{respuesta}=1)}{\text{total de estudiantes}} \times 100 $$

Este cálculo se aplicó tanto a las categorías agregadas (violencia verbal, física y sexual) como a cada ítem específico, con el fin de evaluar patrones de cambio particulares.

#### (c) Visualizaciones
Para evaluar comparativamente ambas encuestas, se elaboraron dos visualizaciones principales:

- Gráfico comparativo por tipo de violencia, que permite identificar diferencias generales en la prevalencia de violencia verbal, física y sexual entre los años 2017 y 2023.

- Gráfico comparativo por ítem de vulneración, que muestra las variaciones específicas en los códigos PA, PB, PC y PD, facilitando la identificación de aumentos o disminuciones en vulneraciones concretas.

Estas visualizaciones apoyan el análisis descriptivo al evidenciar tendencias temporales, variaciones en la magnitud de los reportes y diferencias entre las distintas categorías e indicadores evaluados.

#### **- Modelo Validación Estadística**

```python
df = df_inicial.copy()

columnas_violencia_verbal = ['PB_1', 'PC_3']
columnas_violencia_fisica = ["PA_5", "PA_6", "PA_7", "PB_2", "PC_1", "PC_2", "PC_5"]
columnas_violencia_sexual = ["PD_3", "PD_4", "PD_5", "PD_6", "PD_7"]

columnas_vulneraciones = (
    columnas_violencia_fisica +
    columnas_violencia_verbal +
    columnas_violencia_sexual
)

# 1) Crear variable binaria: ¿sufrió al menos una vulneración?
df['Sufrio_Alguna'] = (df[columnas_vulneraciones] == 1).any(axis=1).astype(int)

# 2) Tabla de contingencia Año x Sufrió_Alguna
contingencia = pd.crosstab(df['Año'], df['Sufrio_Alguna'])

print("\nTabla de contingencia observada:\n")
print(contingencia)

# 3) Test Chi-Cuadrado de independencia
chi2, p, dof, expected = stats.chi2_contingency(contingencia)

print("\n\n --- Resultado Test Chi-Cuadrado (2017 vs 2023) ---")
print(f"Estadístico Chi2: {chi2:.4f}")
print(f"Grados de libertad: {dof}")
print(f"P-value: {p:.4e}")

if p < 0.05:
    print("\n\nCONCLUSIÓN: La diferencia entre 2017 y 2023 ES estadísticamente significativa.")
else:
    print("\n\nCONCLUSIÓN: No hay evidencia estadística suficiente para afirmar que hubo un cambio.")

# 4) Visualización de la prevalencia con el p-value en el título
prop = (
    df.groupby('Año')['Sufrio_Alguna']
      .mean()
      .reset_index()
)

plt.figure(figsize=(8, 8))
sns.barplot(
    data=prop,
    x='Año',
    y='Sufrio_Alguna',
    width=0.5,
    hue='Año',
    errorbar=('ci', 95),
    capsize=.1,
    palette={2017: '#cdcdc2', 2023: '#d39aa7'},
    legend=False
)

plt.title(f'\n\n Prevalencia general de violencia\n(p = {p:.1e})\n', fontsize=18, fontweight='bold')
plt.xlabel('\nAño de encuesta\n', fontsize=12, color='grey')
plt.ylabel('\nProporción de estudiantes afectados\n', fontsize=12, color='grey')
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

Los resultados del test Chi-cuadrado indican que la prevalencia de estudiantes que reportan haber sufrido alguna vulneración es significativamente distinta entre 2017 y 2023 (χ² = 42.89, p < 0.000001).
En términos descriptivos, la proporción aumenta desde aproximadamente 76% en 2017 a 79–80% en 2023, lo que representa un incremento moderado pero consistente.

Este resultado, junto con la evidencia de aumentos por tipo de violencia observada en los análisis previos, sugiere que las vulneraciones reportadas han aumentado de manera real y sistemática durante el período analizado.
El hallazgo podría estar asociado a un aumento efectivo en experiencias de violencia, a una mayor disposición de los estudiantes a reportarlas, o a cambios estructurales en el contexto social y escolar entre ambos años.

## **¿Qué podría salir mal?**

### ***1. La encuesta no representa a todos los niños, niñas y adolescentes de Chile***

Si bien los datos entregan una muestra valiosa, la encuesta no incluye a toda la población infantil y adolescente del país. Esto significa que los resultados reflejan solo a quienes participaron en el estudio y no necesariamente a la realidad completa de Chile. Por lo tanto, las conclusiones no pueden generalizarse al 100% y deben interpretarse como una aproximación parcial.

### ***2. Concentración territorial y sesgo por comuna***

La cantidad de niños, niñas y adolescentes encuestados difiere ampliamente entre comunas, lo que genera un desequilibrio en la base de datos. Algunas zonas cuentan con un número mucho mayor de participantes que otras, lo que puede producir una apariencia engañosa de mayor ocurrencia de violencia en ciertos territorios.
Debido a estas diferencias en el tamaño muestral, no es posible comparar comunas de manera directa, ya que un mayor número de casos registrados podría deberse únicamente a que se encuestó a más personas y no a que exista realmente una frecuencia más alta de violencia. Esta desigualdad limita la representatividad y dificulta obtener conclusiones territoriales precisas.

Con el fin de evaluar la representatividad territorial de los datos, se elabora un mapa de calor que muestra la cantidad de estudiantes encuestados en cada comuna de la Región Metropolitana. Este gráfico permite identificar de manera inmediata el desbalance en la distribución territorial de las observaciones, evidenciando que el número de participantes varía significativamente entre comunas.

```python
# Filtrar solo Región Metropolitana
df_rm = df_inicial[df_inicial['Region_colegio'] == 'Metropolitana'].copy()
comuna_rm = df_rm['Comuna_colegio'].value_counts().reset_index()
comuna_rm.columns = ['Comuna', 'N_estudiantes']

# Crear matriz para el heatmap
heatmap_df = comuna_rm.set_index('Comuna')[['N_estudiantes']]

# Graficar heatmap
plt.figure(figsize=(6, max(6, len(heatmap_df) * 0.25)))

sns.heatmap(
    heatmap_df,
    cmap='Blues',
    annot=True,
    fmt='d',
    linewidths=.5,
    cbar_kws={'label': 'Cantidad de estudiantes encuestados'}
)

plt.title("Distribución de estudiantes encuestados por comuna\n(Región Metropolitana)", 
          fontsize=14, fontweight='bold')
plt.xlabel("")
plt.ylabel("Comuna")
plt.tight_layout()
plt.show()
```
